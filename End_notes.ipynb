# Presentation Video Guidelines - Introduction to Computer Programming

Creating a presentation video with your camera on can be an effective way to engage your audience. This guide will walk you through the steps to create a professional presentation video.

### **What is to be expected?**

- Use the English language for the video explanation.
- Provide detailed explanations for any five selected problems.
- Prepare the video presentation after completing the Colab file while covering all the questions.
- During the explanation, cover the approach, algorithm, and logic used.
- Present technical information in a viewer-friendly manner.
- Ensure the video explanation duration is between 15 to 25 minutes.
- Provide a step-by-step breakdown of the thought process and approach.

### **Choose the Right Platform**

There are several platforms available that allow you to create presentation videos with your camera on. Here are some of the best:

1. **Zoom**
    - **Features**: Virtual backgrounds, recording options, screen sharing, participant interaction.
    - **Link**: [Zoom](https://zoom.us/)
    - **Usage**: Zoom is ideal for live presentations that you want to record. You can use virtual backgrounds and interact with participants in real time.
    - **Platform Setup Link**: [Click Here](https://youtu.be/2csJz_Wo55g?si=tl5z1uLDOu6rxyFS)
2. **OBS Studio (Optional)**
    - **Features**: Free and open-source, advanced recording and streaming, customizable layouts.
    - **Link**: [OBS Studio](https://obsproject.com/)
    - **Usage**: OBS Studio is perfect for **more advanced users** who want complete control over their recording setup. It allows for a high degree of customization and is great for live streaming as well.
    - **Platform Setup Link**: [Click Here](https://youtu.be/yCMdj7B8WiA?si=QBj-f62pW7nh4fVw)

**If you prefer using other platforms, feel free to do so, but please ensure you strictly adhere to the guidelines outlined in the 'What is Expected from You' section.**

### **Set Up Your Recording Space**

1. **Lighting**: Ensure your face is well-lit.
2. **Background**: Choose a clean, uncluttered background. Consider using a virtual background if necessary.
3. **Ensure Clarity**: Make sure your voice is audible, clear, and fluent. Avoid background noise. Speak at a deliberate and steady pace.
4. **Camera Position**: Position your camera at eye level for a more natural and engaging appearance.

### **Record Your Presentation**

1. **Test Your Setup**: Do a test recording to ensure good audio and video quality.
2. **Start Recording**: Begin recording, following your outline and script. Speak clearly and at a steady pace.
3. **Engage with the Camera**: Look directly at the camera to engage in the context.

### **Edit Your Video (If Necessary)**

1. **Trim and Cut**: Remove any mistakes or pauses.
2. **Export and Share**: Export your video in a high-quality format and share it with your audience. Alternatively, you can upload the video to your Google Drive and share the link with **'Anyone with View Access'** permissions.
