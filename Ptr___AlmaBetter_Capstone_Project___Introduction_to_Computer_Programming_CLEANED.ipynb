### Problem 1: Score Board Operator

An operator working behind the scoreboard of an inter-cohort AlmaBetter cricket tournament is responsible for updating the scores and points of each team. However, the operator is currently facing difficulty calculating the total points earned by teams.

According to the tournament's rules, teams are awarded the following points based on the outcome of a match:
- **Wins:** 3 points
- **Draws:** 1 point
- **Losses:** 0 points

**Scenario:**
Team London has played **8 matches** in this tournament. They won **4 matches**, lost **3 matches** and **drew 1**.

The operator is in need of assistance to calculate the total number of points earned by Team London. As a Python expert adept with knowledge of integers, floats and booleans, how would you approach this situation?

**What would your approach be?**

#### Solution: Step 1 - Define your fixed values

```python
# Define your fixed values:
point_per_win = 3
point_per_loss = 0
point_per_drew = 1

total_matches = 8
total_wins = 4
total_loses = 3
total_drew = 1
```

#### Solution: Step 2 - Calculate the total points

```python
# Calculate the total points gained by Team London
total_points = (total_wins * point_per_win) + (total_loses * point_per_loss) + (total_drew * point_per_drew)

print(f"Total points for Team London: {total_points}")
```

### Problem 2: Nutrition Bar Ingredient Chart

Imagine you are a data analyst at a nutrition bar manufacturing company. Your department head approaches you with a question.

The company produces a nutrition bar that contains:
- **45g of raisins**
- **65g of almonds**
- **30g of apricots**

Your department head asks: "What percentage of each ingredient is in the nutrition bar?"

Equipped with the knowledge of int, float and booleans in Python, how would you approach this situation?

#### Solution: Calculate percentages of each ingredient

```python
# Define ingredient quantities
raisins = 45
almonds = 65
apricots = 30

# Calculate total quantity
total_quantity = raisins + almonds + apricots

# Calculate percentages
perc_raisins = (raisins / total_quantity) * 100
perc_almonds = (almonds / total_quantity) * 100
perc_apricots = (apricots / total_quantity) * 100

# Display results
print(f"Total quantity: {total_quantity}g")
print(f"\nIngredient Percentages:")
print(f"Raisins: {perc_raisins:.2f}%")
print(f"Almonds: {perc_almonds:.2f}%")
print(f"Apricots: {perc_apricots:.2f}%")
```

### Problem 3: Measuring the Accuracy of a Disease Detection Model (Optional)

If you're not impressed with these use cases yet, understand that you're just getting started with Python. Going forward as a data scientist, you'll be using what you've learned to design machine learning models.

Let's say you have created a machine learning model to detect diseases in patients based on their symptoms and medical history. The model has been tested on a dataset of **100 patients**, including those with various diseases.

**Model Performance Results:**

The model's performance was evaluated in three rounds:
- **Round 1 (Diabetes):** The model correctly **detected diabetes in 25 out of 30 patients** with the disease.
- **Round 2 (Heart Disease):** The model correctly detected **heart disease in 35 out of 45 patients** with the disease.
- **Round 3 (Cancer):** The model correctly **detected cancer in 20 out of 25 patients** with the disease.

The model's accuracy is defined as the percentage of correctly detected cases out of the total number of patients.

**Task:** Assess how accurate your model is across these 3 diseases.

#### Solution: Store patient data and calculate accuracy

```python
# Store the number of patients for each disease
no_heart_disease = 45
no_diabetes = 30
no_cancer = 25

# Store the number of correct predictions for each disease
correct_heart_disease = 35
correct_diabetes = 25
correct_cancer = 20

# Calculate accuracy for each disease
perc_heart_acc = correct_heart_disease / no_heart_disease
perc_diabetes_acc = correct_diabetes / no_diabetes
perc_cancer_acc = correct_cancer / no_cancer

# Calculate overall accuracy
total_correct = correct_heart_disease + correct_diabetes + correct_cancer
total_patients = no_heart_disease + no_diabetes + no_cancer
perc_overall_acc = total_correct / total_patients

# Display results
print("Model Accuracy Results:")
print("=" * 50)
print(f"\nDisease-Specific Accuracy:")
print(f"Heart Disease:  {perc_heart_acc:.2%} ({correct_heart_disease}/{no_heart_disease})")
print(f"Diabetes:       {perc_diabetes_acc:.2%} ({correct_diabetes}/{no_diabetes})")
print(f"Cancer:         {perc_cancer_acc:.2%} ({correct_cancer}/{no_cancer})")
print(f"\nOverall Accuracy: {perc_overall_acc:.2%} ({total_correct}/{total_patients})")
print("=" * 50)
```

## Data Types in Python

### Solving Algebra the Cool Way 😎

Python can be used to solve algebraic equations and mathematical problems efficiently. Here are examples of key concepts:

#### Example: Working with Different Data Types

```python
# Integer operations
a = 10
b = 3
print(f"Integer Division: {a} // {b} = {a // b}")
print(f"Modulo: {a} % {b} = {a % b}")
print(f"Power: {a} ** {b} = {a ** b}")

# Float operations
x = 10.5
y = 3.2
print(f"\nFloat Addition: {x} + {y} = {x + y}")
print(f"Float Division: {x} / {y} = {x / y}")

# Boolean operations
is_valid = True
is_empty = False
print(f"\nBoolean AND: {is_valid} and {is_empty} = {is_valid and is_empty}")
print(f"Boolean OR: {is_valid} or {is_empty} = {is_valid or is_empty}")
print(f"Boolean NOT: not {is_valid} = {not is_valid}")
```
